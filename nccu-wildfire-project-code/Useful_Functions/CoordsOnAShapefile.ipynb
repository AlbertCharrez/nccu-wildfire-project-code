```python
import geopandas as gpd
import matplotlib.pyplot as plt
from shapely.geometry import Point
```

```python
# Load existing shapefile (optional background)
shapefile = gpd.read_file("/hpc/home/a1/FishLakeFireData/Shapefile/Perimeters.shp")

# Create a point coordinate
point = Point(-111.9894, 38.5835)  # longitude, latitude

# Create GeoDataFrame for the point
point_gdf = gpd.GeoDataFrame(
    {"name": ["My Point"]},
    geometry=[point],
    crs="EPSG:4326"
)

# Plot
fig, ax = plt.subplots(figsize=(10, 8))
shapefile.plot(ax=ax, color="lightgray", edgecolor="black")
point_gdf.plot(ax=ax, color="red", markersize=50, marker="o", label="My Point")

plt.legend()
plt.title("Point on Shapefile")
plt.show()
```

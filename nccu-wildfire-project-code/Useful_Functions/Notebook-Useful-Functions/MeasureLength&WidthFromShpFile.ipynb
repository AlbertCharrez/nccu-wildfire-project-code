```python
import geopandas as gpd

# Load shapefile
gdf = gpd.read_file("/hpc/home/a1/FishLakeFireData/Shapefile/Perimeters.shp")

# Reproject to a meter-based CRS
gdf = gdf.to_crs("EPSG:3857")

# Bounding box
minx, miny, maxx, maxy = gdf.total_bounds

# Dimensions in meters
width_m = maxx - minx
height_m = maxy - miny

# Convert to kilometers
width_km = width_m / 1000
height_km = height_m / 1000

print(f"Width: {width_km:.2f} km")
print(f"Height: {height_km:.2f} km")
```

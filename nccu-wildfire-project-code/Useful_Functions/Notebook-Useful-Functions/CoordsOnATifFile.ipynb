```python
import rasterio
import matplotlib.pyplot as plt

coord_x = -111.9894
coord_y = 38.5835

with rasterio.open("/hpc/home/a1/FishLakeFireData/IncMergeFiles/hv_1.tif") as src:
    img = src.read(1)

    plt.figure(figsize=(10,8))
    plt.imshow(
        img,
        extent=[src.bounds.left, src.bounds.right,
                src.bounds.bottom, src.bounds.top],
        cmap="gray"
    )

    plt.scatter(coord_x, coord_y, s=80, marker='x')
    plt.text(coord_x, coord_y, " Target Coord", fontsize=12)

    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.show()
```

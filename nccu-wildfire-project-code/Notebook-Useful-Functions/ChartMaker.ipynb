```python
import matplotlib.pyplot as plt

minsize = [50, 100, 150, 200]

accuracy = [88.95, 89.53, 89.51, 89.61]
precision = [83.45, 84.36, 84.27, 84.44]
recall = [92.04, 92.28, 92.35, 92.36]
f1 = [87.53, 88.14, 88.13, 88.22]

plt.plot(minsize, accuracy, marker='o', label='Accuracy')
plt.plot(minsize, precision, marker='o', label='Precision')
plt.plot(minsize, recall, marker='o', label='Recall')
plt.plot(minsize, f1, marker='o', label='F1-score')

plt.xlabel('Minsize')
plt.ylabel('Score')
plt.title('Performance Metrics vs Minsize (σ = 8)')
plt.ylim(80, 95)
plt.legend(loc='lower right')
plt.grid()
plt.savefig("/hpc/home/a1/sigma2 scores")

plt.show()
```

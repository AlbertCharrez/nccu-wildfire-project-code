```python
import geopandas as gpd
from shapely.geometry import shape

# Load fire perimeter shapefile (e.g., from NIFC or GeoMAC)
gdf = gpd.read_file("/hpc/home/a1/FishLakeFireData/Shapefile/Perimeters.shp")

# Reproject to equal-area CRS for accurate centroid
gdf = gdf.to_crs("EPSG:5070")  # Albers Equal Area

# Get centroid
centroid = gdf.geometry.centroid

# Convert back to lat/lon (WGS84)
gdf_wgs84 = gdf.to_crs("EPSG:4326")

gdf_projected = gdf_wgs84.to_crs(epsg=3857)
centroid_projected = gdf_projected.geometry.centroid
centroid_wgs84 = centroid_projected.to_crs(epsg=4326)

centroid_wgs84 = gdf_wgs84.geometry.centroid

lat = centroid_wgs84.y.values[0]
lon = centroid_wgs84.x.values[0]

print(f"Perimeter Center: {lat:.4f}, {lon:.4f}")

# Use the equal-area projection (meters)
gdf_proj = gdf.to_crs("EPSG:5070")

# Get bounding box: (minx, miny, maxx, maxy)
minx, miny, maxx, maxy = gdf_proj.total_bounds

width_m  = maxx - minx        # East-West extent (meters)
height_m = maxy - miny        # North-South extent (meters)

print(f"Width  (E-W): {width_m:.1f} m  ({width_m/1000:.2f} km)")
print(f"Height (N-S): {height_m:.1f} m  ({height_m/1000:.2f} km)")
```

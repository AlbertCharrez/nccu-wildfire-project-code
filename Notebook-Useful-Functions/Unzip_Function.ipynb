```python
import zipfile
import os

zip_filename = '/hpc/home/a1/WildfireProject/nccu-wildfire-project-code/data-processing/Processing-Tools.zip'

if not os.path.exists(zip_filename):
    print(f"Error: {zip_filename} not found. Please upload the file.")
else:
    # Create a ZipFile object
    with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
        # Extract all the contents into the current directory
        os.chdir('/hpc/home/a1/WildfireProject/nccu-wildfire-project-code/data-processing')
        zip_ref.extractall('.')
    print(f"Successfully extracted {zip_filename}")
```
